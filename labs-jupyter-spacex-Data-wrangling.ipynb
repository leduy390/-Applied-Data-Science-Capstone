# **SpaceX Falcon 9 First Stage Landing Prediction**
## **Lab 3: Exploratory Data Analysis & Data Wrangling**

In this lab, we perform Exploratory Data Analysis (EDA) and Data Wrangling on the Falcon 9 launch dataset (`dataset_part_1.csv`) to convert landing outcomes into supervised binary classification labels (`Class = 1` for successful landing, `Class = 0` for unsuccessful/no landing) and export the processed dataset to `dataset_part_2.csv`.

### **Objectives**
* Perform Exploratory Data Analysis (EDA) to inspect data types, missing values, and feature distributions.
* Determine launch counts per launch site and orbit type.
* Analyze the unique landing outcomes (`Outcome`) to define positive and negative classes.
* Engineer the binary target variable `Class` (`1` = success, `0` = failure) and compute the overall success rate.
* Export the wrangled dataset to `dataset_part_2.csv`.

### **1. Import Libraries & Load Dataset**

```python
import pandas as pd
import numpy as np

# Load dataset collected from SpaceX REST API
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv")
print("Dataset Shape:", df.shape)
df.head(10)
```

```python
# Calculate percentage of missing values in each attribute
print("Percentage of Missing Values per Column:")
print((df.isnull().sum() / len(df)) * 100)

print("\nColumn Data Types:")
print(df.dtypes)
```

### **2. Task 1 & Task 2: Calculate Launch Counts by Launch Site and Orbit**

```python
# Task 1: Calculate the number of launches on each site
print("Launches per Launch Site:")
print(df['LaunchSite'].value_counts())

# Task 2: Calculate the number and occurrence of each orbit
print("\nLaunches per Orbit Type:")
print(df['Orbit'].value_counts())
```

### **3. Task 3: Calculate the Number and Occurrence of Mission Outcomes per Orbit Type**

```python
# Count landing outcomes across all 90 launches
landing_outcomes = df['Outcome'].value_counts()
print("Landing Outcomes Distribution:")
print(landing_outcomes)

print("\nIndexed Landing Outcomes:")
for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)
```

### **4. Task 4: Create a Landing Outcome Label (`Class`) from `Outcome` Column**

```python
# Define set of unsuccessful/non-landed outcomes (indices 1, 3, 5, 6, 7)
bad_outcomes = set(landing_outcomes.keys()[[1, 3, 5, 6, 7]])
print("Unsuccessful Landing Outcomes (Class = 0):", bad_outcomes)

# Create binary Class list: 0 if outcome in bad_outcomes, else 1
landing_class = [0 if outcome in bad_outcomes else 1 for outcome in df['Outcome']]
df['Class'] = landing_class

print("\nFirst 8 Class labels:", df[['Class']].head(8)['Class'].tolist())
print(f"Overall First-Stage Landing Success Rate: {df['Class'].mean():.4f} ({df['Class'].mean()*100:.2f}%)")
```

```python
# Verify final DataFrame structure and export to dataset_part_2.csv
df.to_csv("dataset_part_2.csv", index=False)
print("Successfully exported dataset_part_2.csv (90 rows, 18 columns).")
df[['FlightNumber', 'Date', 'PayloadMass', 'Orbit', 'LaunchSite', 'Outcome', 'Class']].tail(5)
```
